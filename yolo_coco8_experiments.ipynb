## Modelos testados: YOLOv10 e YOLO26 nas variantes Nano(n) e Small(s)
### Baseline: YOLOv8 nas variantes Nano(n) e Small(s)

A YOLOv10 foi escolhida por apresentar treinamento livre de NMS e uma arquitetura orientada pela eficiência e precisão, entregando desempenho e latência de ponta. A YOLO26 foi escolhida por ser uma evolução da YOLOv10.

```python
models = [
    ("yolov8n", "yolov8n.pt"),
    ("yolov8s", "yolov8s.pt"),
    ("yolov10n", "yolov10n.pt"),
    ("yolov10s", "yolov10s.pt"),
    ("yolo26n", "yolo26n.pt"),
    ("yolo26s", "yolo26s.pt"),
]

results = {}
```

### Treinando com o dataset COCO8 (Detecção de Objetos), com 50 épocas

```python
for name, weights in models:
  model = YOLO(weights)
  results[name] = model.train(
      data="coco8.yaml",
      epochs=50,
      imgsz=640,
      project="coco8",
      name=name,
      exist_ok=True,
      save=True,
      plots=True
  )
```
